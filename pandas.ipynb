```python
series = pd.Series({"Day 1": 9.4, "Day 2": 2.5, "Day 3": 5.0})

# filter by key
series.loc["Day 3"] += 3.2

# filter by values
filter_by_values = series[series >= 5.0]
print(filter_by_values)
# print(series) 
```
