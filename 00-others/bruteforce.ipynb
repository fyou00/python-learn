```python
import pandas as pd
import random

def apaaja(exp, n):
    return [exp for _ in range(n)]

data = {"Weight": apaaja(random.randrange(1,9), 4), "Profit": apaaja(random.randrange(10,90), 4)}

capacity = 14

df = pd.DataFrame(data)
print(df)

```

```python
def apaaja(exp, n):
    return [exp() for _ in range(n)]

data = {
    "Weight": apaaja(lambda: random.randrange(1, 9), 4),
    "Profit": apaaja(lambda: random.randrange(10, 90), 4)
}

df = pd.DataFrame(data)
print(df)
```
