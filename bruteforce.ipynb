```python
import pandas as pd
import random
    
def getWeight(n):
    return [random.randrange(1,9) for _ in range(n)]

def getProfit(n):
    return [random.randrange(10,90,10) for _ in range(n)]

data = {"Weight": getWeight(4), "Profit": getProfit(4)}

df = pd.DataFrame(data)
print(df)
```
