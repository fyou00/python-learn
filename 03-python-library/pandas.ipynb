```python
series = pd.Series({"Day 1": 9.4, "Day 2": 2.5, "Day 3": 5.0})

# filter by key
series.loc["Day 3"] += 3.2

# filter by values
print(series[series >= 5.0])
# print(series) 
```

```python
# DataFrame -> struktur data tabel dengan kolom dan baris

data = {"Name": ["Furina", "Escoffier", "Diluc", "Amber"], "Age": [400, 22, 30, 20]}

df = pd.DataFrame(data, index=["Char 1", "Char 2", "Char 3", "Char 4"])

# Add a new column
df["Element"] = ["Hydro", "Cryo", "Pyro", "Pyro"]

# print(df)

# loc adalah location berdasarkan label, berikan aku data char yg berada pada index 2
print(df.iloc[2])
```
